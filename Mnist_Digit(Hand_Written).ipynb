```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras import Model
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
```

## Load MNIST dataset (handwritten digit : 0-9):

```python
# Load MNIST dataset (handwritten digit : 0-9):
 (x_train, y_train), (x_test, y_test) = mnist.load_data()
```

# Normalisze the data (scale pixel values berween 0 to 1)

```python
# Normalisze the data (scale pixel values berween 0 to 1)
x_train = x_train / 255.0
x_test = x_test / 255.0
```

# One-hot encode the labels (0-9)


```python
# One-hot encode the labels (0-9)
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
```

# Define a Multi-Layer Preceptron model

```python
# Define a Multi-Layer Preceptron model
model = Sequential([
    Flatten(input_shape=(28, 28)),               # Flatten 28x28 images into 784 vector
    Dense(128, activation='relu'),               # Hidden Layer 1 with 128 neurons
    Dense(64, activation='relu'),                # Hidden Layer 2 with 64 neurons
    Dense(10, activation='softmax')              # Output Layer (10 clsses)
])
```

```python
# compile the model
model.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])
```

```python
from re import VERBOSE
# Train
model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.1)
```

```python
# Evaluate the model on test data
test_loss, test_acc = model.evaluate(x_test, y_test, verbose = 0)
print(f"Test accuracy: {test_acc: .4f}")
```
